```python
#Functions Practice 2 - for loops

#Write a function where given the below array, it adds 1 to every interger and returns another array of those integers
#Example input: 
array2 = [1,2,3,4]
#Example output: [2,3,4,5]

def addone(array):
    ret = []
    for x in array:
        ret.append(x+1)
    return ret

addone(array2)
```

```python
#Functions Practice 3 - if/else statements

#Write a function where given the below array, only returns two arrays: one with the even numbers and one with the odd
#Example input: 
array3 = [2,3,4,5,6]
#Exmaple output: [2,4,6],[3,5]

def findevens(array):
    array_evens = []
    array_odds = []
    for x in array:
        if x % 2 == 0:
            array_evens.append(x)
        else:
            array_odds.append(x)
    return (array_evens, array_odds)

findevens(array3)
```

```python
#Functions Practice 4 - nested loops

#Write a functions where given the two arrays below, it pairs the words together and returns them in their pairs
#Example input:
array4a = ["Sarah", "Zach", "Sam"]
array4b = ["Rifampin", "Isoniazid", "Pyrazinamide", "Ethambutol"]
#Example output:
# Sarah Rifampin
# Sarah Isoniazid
# Sarah Pyrazinamide
# Sarah Ethambutol
# Zach Rifampin ...
# etc etc etc
#Hint: I would simply use "print("xyz")" within your function instead of creating an empty array
#This is an example where you don't need a return statement if you use print

def pairs (array1, array2):
    for x in array1:
        for y in array2:
            print(x, y)

pairs(array4a,array4b)
```
