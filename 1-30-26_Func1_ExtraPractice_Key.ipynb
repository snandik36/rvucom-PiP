Your job is to create a function that acts as a mini triage system. You are given a list of patients. Based on the following criteria, decide what level the patient is - return a list of patients that are level 1 or level 2.

Level 1 (Critical) if:
    SBP < 90 OR
    heart rate > 130 AND complaint is chest pain or shortness of breath

Level 2 if:
    age ≥ 65 AND
    heart rate > 110

Level 3 if:
    heart rate between 100–110

Level 4 if:
    complaint is musculoskeletal pain
    
Level 5 otherwise

```python
patients = [
    {
        "name": "John Smith",
        "age": 72,
        "heart_rate": 132,
        "sbp": 85,
        "complaint": "chest pain"
    },
    {
        "name": "Maria Lopez",
        "age": 68,
        "heart_rate": 115,
        "sbp": 140,
        "complaint": "dizziness"
    },
    {
        "name": "Kevin Patel",
        "age": 45,
        "heart_rate": 140,
        "sbp": 110,
        "complaint": "shortness of breath"
    },
    {
        "name": "Aisha Khan",
        "age": 30,
        "heart_rate": 102,
        "sbp": 118,
        "complaint": "abdominal pain"
    },
    {
        "name": "Robert Chen",
        "age": 80,
        "heart_rate": 108,
        "sbp": 95,
        "complaint": "weakness"
    },
    {
        "name": "Emily Johnson",
        "age": 22,
        "heart_rate": 78,
        "sbp": 122,
        "complaint": "ankle pain"
    },
    {
        "name": "Samuel Green",
        "age": 67,
        "heart_rate": 112,
        "sbp": 135,
        "complaint": "fatigue"
    },
    {
        "name": "Lina Morales",
        "age": 55,
        "heart_rate": 90,
        "sbp": 88,
        "complaint": "chest pain"
    }
]
```

```python
def is_level_1(patient):
    critical_sbp = patient["sbp"] < 90
    critical_hr_and_complaint = (
        patient["heart_rate"] > 130 and
        patient["complaint"] in ["chest pain", "shortness of breath"]
    )
    return critical_sbp or critical_hr_and_complaint


def is_level_2(patient):
    return patient["age"] >= 65 and patient["heart_rate"] > 110


def get_level_1_or_2_patients(patients):
    high_priority = []

    for patient in patients:
        if is_level_1(patient) or is_level_2(patient):
            high_priority.append(patient)

    return high_priority
```

```python
result = get_level_1_or_2_patients(patients)
for p in result:
    print(p["name"])
```

Here's another way to do it!

```python
def get_level_1_or_2_patients(patients):
    high_priority = []

    for patient in patients:
        # Level 1 conditions
        level_1 = (
            patient["sbp"] < 90 or
            (
                patient["heart_rate"] > 130 and
                patient["complaint"] in ["chest pain", "shortness of breath"]
            )
        )

        # Level 2 conditions
        level_2 = (
            patient["age"] >= 65 and
            patient["heart_rate"] > 110
        )

        if level_1 or level_2:
            high_priority.append(patient["name"])

    return high_priority

get_level_1_or_2_patients(patients)
```
