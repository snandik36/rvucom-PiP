```python
#Try to re-write the factorial function we did in class

def factorial(n):
    if (n==1 or n==0):  
        return 1
    else: 
        return (n * factorial(n - 1))

factorial(3)
# should return 6
```

```python
#Extra recursion practice:
# Create a recusrive function for computing the nth fibonacci number
# Fibonacci is a sequence of numbers that follows this function: F(n) = F(n-1) + F(n-2)
# The sequence is 0, 1, 1, 2, 3, 5, 8, etc...
# This function does use "elif" - which is an addition to if/else that basically adds another clause to your statement

def fibonacci(n):
    if n < 0:
        print("Incorrect input")
    elif n == 0:
        return 0
    elif n == 1 or n == 2:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
    
print(fibonacci(10))
# should return 55
```
